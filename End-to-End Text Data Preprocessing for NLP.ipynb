```python
import re


text = """Hello!!! This is a SAMPLE text for NLP preprocessing 😊😊.
Visit our website at https://www.example.com for more details.

I can't believe how AMAZING this product is!!! 😍😍
I've been using it for 2 years, and it's still working perfectly.

Contact us at support@example.com or call +91-9876543210.
Follow us on Twitter @example_support and Instagram #ExampleBrand.

Data Science, Machine Learning, and AI are the future!!!
But wait... is this REALLY true??? 🤔🤔

The quick brown fox jumps over the lazy dog.
THE QUICK BROWN FOX JUMPS OVER THE LAZY DOG!!!

Breaking News: India’s GDP growth reached 7.6% in 2024!!!
Experts say: “This is a remarkable achievement.”

I love running, runs, and ran every morning.
Better results are achieved when you are consistent.

Stopwords like the, is, and, of appear frequently in text data.
Removing them helps models focus on meaningful information.

Check this out 👉 https://bit.ly/3xyzAbC
Email me at test.user123@gmail.com ASAP!!!

#NLP #DataScience #AI #MachineLearning"""



# 1. Remove URLs (http, https, www, bit.ly, etc.)
clean_step3 = re.sub(r"https?://\S+|www\.\S+", "", text)

# 2. Remove Email addresses
clean_step3 = re.sub(r"\b[\w\.-]+@[\w\.-]+\.\w+\b", "", clean_step3)

# 3. Remove Phone numbers (handles +91-9876543210 and international formats)
clean_step3 = re.sub(
    r"(\+?\d{1,3}[-.\s]?)?\(?\d{3}\)?[-.\s]?\d{3}[-.\s]?\d{4}|\+?\d{1,4}[-.\s]?\d{10}\b",
    "",
    clean_step3,
)

# 4. Remove Mentions (@username)
clean_step3 = re.sub(r"@\w+", "", clean_step3)

# 5. Remove Hashtags (#topic)
clean_step3 = re.sub(r"#\w+", "", clean_step3)




# 1. Convert all text to lowercase
normalized = clean_step3.lower()


# 3. Remove Numbers (e.g., 2, 7, 6, 2024)
normalized = re.sub(r"\d+", "", normalized)

# 4. Remove Punctuation and Special Characters (!, ?, ., %, ", ', etc.)
# Custom regex handles standard punctuation plus curly quotes/apostrophes
normalized = re.sub(r"[^\w\s]|_", "", normalized)

# 5. Remove Extra Whitespaces, newlines, and tabs
normalized = re.sub(r"\s+", " ", normalized).strip()



print("--- TEXT AFTER NOISE REMOVAL (STEP 3) ---")
print(clean_step3.strip())

print("\n--- FINAL NORMALIZED TEXT (STEP 2) ---")
print(normalized)
```

```python
from nltk.tokenize import word_tokenize
```

```python
tokens = word_tokenize(normalized)
```

```python
tokens[:20]
```

```python
import nltk
from nltk.corpus import stopwords
```

```python
stop_words = set(stopwords.words('english'))
```

```python
print(stop_words)
```

```python
filtered_tokens = [
    word for word in tokens
    if word not in stop_words
]
```

```python
print(tokens) ,len(tokens)
```

```python
print(filtered_tokens) , len(filtered_tokens)
```

```python
from nltk.stem import PorterStemmer ,WordNetLemmatizer
```

```python
stemmer = PorterStemmer()
```

```python
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]
```

```python
stemmed_words
```

```python
lemma = WordNetLemmatizer()
```

```python
lemma_words = [lemma.lemmatize(word) for word in stemmed_words]
```

```python
lemma_words
```

```python
clean_text = ""
```

```python
for i in lemma_words:
    clean_text = clean_text+" "+i
```
